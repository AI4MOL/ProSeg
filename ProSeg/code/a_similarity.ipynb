```python
import numpy as np
from joblib import Parallel, delayed
from tqdm import tqdm
import numpy as np
import glob
# Get a list of all .h5 files in the directory
h5_files = glob.glob('/data/DP/testing/*.h5')

# Load each .h5 file
h5_data = {}
for file in h5_files:
    with h5py.File(file, 'r') as f:
        h5_data[file] = {key: f[key][()] for key in f.keys()}


similar_matrix = np.ones((len(h5_files), 4, 4))
# 定义计算单个切片的相似矩阵
def compute_similarity(one_slice_id, data):
    temp_matrix = np.ones((4, 4))
    for i in range(4):
        for j in range(i + 1, 4):
            temp_matrix[i, j] = distance_one(data[f'label_a{i+1}'][one_slice_id].flatten(),
                                         data[f'label_a{j+1}'][one_slice_id].flatten())
            temp_matrix[j, i] = temp_matrix[i, j]  # 对称赋值
    return temp_matrix

# 初始化计数
count = 0

# 遍历 h5 文件
for file, data in h5_data.items():
    num_slices = data['label_a1'].shape[0]
    
    # 使用 joblib 并行计算所有切片的相似矩阵
    temp_similar_matrices = Parallel(n_jobs=-1)(
        delayed(compute_similarity)(one_slice_id, data) for one_slice_id in tqdm(range(num_slices))
    )

    # 计算均值并存储
    similar_matrix[count, :, :] = np.mean(temp_similar_matrices, axis=0)
    count += 1
np.save('vis/similar_matrix.npy', similar_matrix)
```
